```python
from pathlib import Path
from cfclone.plot.plot_ctdna import plot_ctdna
from cfclone.plot.plot_clones_from_tsv import plot_clones_from_tsv

out_dir = Path('results')
out_dir.mkdir(exist_ok=True, parents=True)
ctdna_file = "data/cfdna.tsv.gz"
clone_cn_file = "data/clone_cn.tsv.gz"
ctdna_plot_file = out_dir.joinpath('ctdnas.png')
clone_cn_plot_file = out_dir.joinpath('clone_cn.png')
plot_ctdna(ctdna_file, plot_file=out_dir.joinpath('rdr.png'), yvar='rdr', title='RDR')
plot_ctdna(ctdna_file, plot_file=out_dir.joinpath('baf.png'), yvar='baf', title='BAF')
plot_clones_from_tsv(clone_cn_file, clone_cn_plot_file)
```

![Alt text](./results/rdr.png)
![Alt text](./results/baf.png)
![Alt text](./results/clone_cn.png)

```python
from cfclone.run import fit

fit(
    in_file=ctdna_file,
    clone_cnv_file=clone_cn_file,
    out_file=out_dir.joinpath('fit.h5'),
    add_normal=True,
    outlier=True,
    num_chains=2,
    num_rounds=2,
    num_threads=2,
    use_clone='8',
    num_bins=300,
)
```

```python
from cfclone.run.postprocess import write_parameter_summaries
from cfclone.plot.plot_fit import plot_fit

write_parameter_summaries(
    in_file=out_dir.joinpath('fit.h5'),
    out_file=out_dir.joinpath('parameter_summaries.tsv'),
)
plot_fit(
    in_file=out_dir.joinpath('parameter_summaries.tsv'),
    out_file=out_dir.joinpath('fit.png')
)
```
